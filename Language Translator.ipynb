# Project Tittle : English-to-Other Language Translator

## Objective:
To build a program that translates English text into other languages like French and Hindi using Hugging Face’s translation tools.

## Steps:

1. Install Required Library
2. Import Translation Pipeline
3. Initialize Translator Model
4. Input English Text
5. Perform Translation
6. Display Translated Text

### Step 1: French Translation (English → French)

```python
from transformers import MarianMTModel, MarianTokenizer

# Step 1: Specify the translation model (English to French)
model_name_fr = 'Helsinki-NLP/opus-mt-en-fr'

# Step 2: Load tokenizer and model
tokenizer_fr = MarianTokenizer.from_pretrained(model_name_fr)
model_fr = MarianMTModel.from_pretrained(model_name_fr)

# Step 3: Input English sentence
text = "Hello, how are you?"

# Step 4: Tokenize and prepare the input
inputs_fr = tokenizer_fr(text, return_tensors="pt", padding=True)

# Step 5: Generate the translation
translated_fr = model_fr.generate(**inputs_fr)

# Step 6: Decode and display the output
output_fr = tokenizer_fr.decode(translated_fr[0], skip_special_tokens=True)
print("Translated (English to French):", output_fr)
```
