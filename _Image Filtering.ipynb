### Step 1: Importing Libraries

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

### Step 2: Load the Image

```python
# Load image 
img = cv2.imread(r"C:\Users\LENOVO\Desktop\padmavathi k\dog.jpg")
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)  
```

### Step 3: Apply Filters

```python
# Grayscale
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
# Blur
blur_img = cv2.GaussianBlur(img_rgb, (15, 15), 0)
```

```python
# Sepia filter
sepia_filter = np.array([[0.272, 0.534, 0.131],
                         [0.349, 0.686, 0.168],
                         [0.393, 0.769, 0.189]])
sepia_img = cv2.transform(img, sepia_filter)
sepia_img = np.clip(sepia_img, 0, 255).astype(np.uint8)
sepia_img = cv2.cvtColor(sepia_img, cv2.COLOR_BGR2RGB)
```

```python
# Edge Detection
edge_img = cv2.Canny(img, 100, 200)
```

### Step 4: Save the Filtered Images

```python
# Save processed images
cv2.imwrite(r"C:\Users\LENOVO\Desktop\padmavathi k\dog_gray.jpg", gray_img)
cv2.imwrite(r"C:\Users\LENOVO\Desktop\padmavathi k\dog_blur.jpg", cv2.cvtColor(blur_img, cv2.COLOR_RGB2BGR))
cv2.imwrite(r"C:\Users\LENOVO\Desktop\padmavathi k\dog_sepia.jpg", cv2.cvtColor(sepia_img, cv2.COLOR_RGB2BGR))
cv2.imwrite(r"C:\Users\LENOVO\Desktop\padmavathi k\dog_edges.jpg", edge_img)
```

### Step 5: Show All Images Using Matplotlib

```python
# Display using matplotlib
plt.figure(figsize=(12, 8))
plt.subplot(2, 3, 1)
plt.imshow(img_rgb)
plt.title("Original")
plt.axis('off')
plt.tight_layout()#It hides the lines and numbers around the image.
plt.show()
```

```python
plt.figure(figsize=(12, 8))
plt.subplot(2, 3, 2)
plt.imshow(gray_img, cmap='gray')
plt.title("Grayscale")
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 8))
plt.subplot(2, 3, 3)
plt.imshow(blur_img)
plt.title("Blur")
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 8))
plt.subplot(2, 3, 4)
plt.imshow(sepia_img)
plt.title("Sepia")
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 8))
plt.subplot(2, 3, 5)
plt.imshow(edge_img, cmap='gray')
plt.title("Edge Detection")
plt.axis('off')
plt.tight_layout()
plt.show()
```
